```python
import numpy as np
import matplotlib.pyplot as plt
import flashy
from flashy.io import TimeSeries1D, DatFile
from flashy.nuc import Tracers1D

%matplotlib inline
Msun = 1.989e33
```

```python
# Read simulation plt files from directory, taking 2ms as step size
sim_data = Tracers1D.read_simulation('./s20/', step_size=2)
```

```python
# Determine mass shells coordinates most suitable for tracers, i.e. crossing the shock and ejected in the neutrino wind
# Given the simulation data (time series), dat file (DatFile object or file path), and desired number of tracers
masses = Tracers1D.calculate_mass_shells(sim_data, 's20.dat', n_shells=20)
print(masses/Msun)
```

```python
# Determine each tracer seed nuclei
# Given the mass coordinates and progenitor file.
# The progenitor file must include an enclosed mass variable ('mass'), and isotopic data (the available isotopes are detected automatically)
mass_fractions, detected_network = Tracers1D.calculate_seeds('s20.0.1d', masses)
print(detected_network)

# Save seeds (one file per tracer) in 'seeds' directory
Tracers1D.save_seeds(detected_network, mass_fractions, './seeds/')
```

```python
# Integrate trajectories
# Given simulation data and mass coordinates
trajectories = Tracers1D.calculate_trajectories(sim_data, masses)

# Save trajectories (one file per tracer) in 'traj' directory
# Trajectories are only saved from when tracer goes below Tmax temperature to end of the simulation (default is 10GK)
Tracers1D.save_trajectories(trajectories, './traj/', Tmax=1e10)
```
